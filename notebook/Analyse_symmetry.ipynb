```python
from pymatgen.core import Structure
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer

# Read POSCAR
structure = Structure.from_file("/Users/aasthapimoli/Comp_Chem/Crsytal_Struct_nd_Sym_part 2/Refined Jupytercodes/Create_struct/POSCAR_CMP_triclinic.vasp")

# Symmetry analysis
sga = SpacegroupAnalyzer(structure, symprec=1e-3, angle_tolerance=5)

# Basic symmetry information
print("=" * 60)
print("Symmetry Analysis")
print("=" * 60)

print(f"Space Group Symbol : {sga.get_space_group_symbol()}")
print(f"Space Group Number : {sga.get_space_group_number()}")
print(f"Crystal System     : {sga.get_crystal_system()}")
print(f"Lattice Type       : {sga.get_lattice_type()}")
print(f"Hall Symbol        : {sga.get_hall()}")

# Symmetrized structure
symm_structure = sga.get_symmetrized_structure()

print("\nEquivalent Atomic Sites:")
for i, eq_sites in enumerate(symm_structure.equivalent_sites):
    print(f"Wyckoff Set {i+1}")
    print(f"  Multiplicity: {len(eq_sites)}")
    print(f"  Species     : {eq_sites[0].species_string}")
    print(f"  Wyckoff     : {symm_structure.wyckoff_symbols[i]}")
    print()

# Conventional and primitive cells
conv_structure = sga.get_conventional_standard_structure()
prim_structure = sga.get_primitive_standard_structure()

print("=" * 60)
print("Cell Information")
print("=" * 60)
print(f"Original atoms     : {len(structure)}")
print(f"Primitive atoms    : {len(prim_structure)}")
print(f"Conventional atoms : {len(conv_structure)}")

# Symmetry operations
symm_ops = sga.get_symmetry_operations()

print("\nNumber of symmetry operations:", len(symm_ops))

for i, op in enumerate(symm_ops[:10]):  # show first 10
    print(f"\nOperation {i+1}")
    print("Rotation:")
    print(op.rotation_matrix)
    print("Translation:")
    print(op.translation_vector)
```
